```python
"""
LITE ВЕРСИЯ эксперимента VSSD для Google Colab (CPU-версия)
Время выполнения: ~45-60 минут на CPU
"""

!pip install torch torchvision matplotlib tqdm -q

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm
import time
import warnings
warnings.filterwarnings('ignore')

# ==================== КОНФИГУРАЦИЯ (CPU ОПТИМИЗИРОВАНА) ====================
class LiteConfig:
    # Данные (уменьшенный набор для CPU)
    dataset = "CIFAR10"  # вместо CIFAR100 для скорости
    num_classes = 10
    img_size = 32  # сохраняем оригинальный размер CIFAR для скорости

    # Обучение (сильно уменьшено для CPU)
    batch_size = 16  # уменьшили для CPU
    num_epochs = 15   # всего 5 эпох для скорости
    learning_rate = 1e-3
    subset_size = 1000  # используем маленькое подмножество данных

    # Модели (упрощенные и легкие)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    # Для воспроизводимости
    seed = 42

config = LiteConfig()
torch.manual_seed(config.seed)

print(f"Устройство: {config.device}")
if torch.cuda.is_available():
    print(f"✓ GPU доступен: {torch.cuda.get_device_name(0)}")
else:
    print("⚠ Используется CPU (будет медленнее)")
    print("  Для ускорения в Colab: Runtime → Change runtime type → GPU")

# ==================== ДАННЫЕ (СУПЕР-LITE) ====================
def get_lite_datasets():
    """Минимальный набор данных"""
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
    ])

    if config.dataset == "CIFAR10":
        print("Загрузка CIFAR-10...")
        full_train = torchvision.datasets.CIFAR10(
            root='./data', train=True, download=True, transform=transform
        )
        full_test = torchvision.datasets.CIFAR10(
            root='./data', train=False, download=True, transform=transform
        )
    else:
        print("Загрузка CIFAR-100...")
        full_train = torchvision.datasets.CIFAR100(
            root='./data', train=True, download=True, transform=transform
        )
        full_test = torchvision.datasets.CIFAR100(
            root='./data', train=False, download=True, transform=transform
        )

    # Берем только подмножество для скорости
    indices = list(range(min(config.subset_size, len(full_train))))
    train_size = int(len(indices) * 0.8)
    train_dataset = Subset(full_train, indices[:train_size])
    test_dataset = Subset(full_test, indices[train_size:train_size + int(len(indices)*0.2)])

    return train_dataset, test_dataset

# ==================== УПРОЩЕННАЯ VSSD (CPU-ОПТИМИЗИРОВАНА) ====================
class LiteVSSD(nn.Module):
    """Минимальная версия VSSD для демонстрации на CPU"""
    def __init__(self, in_channels=3, num_classes=10, dim=32):  # уменьшили dim для CPU
        super().__init__()

        # Простой патч эмбеддинг
        self.patch_embed = nn.Conv2d(in_channels, dim, kernel_size=4, stride=4)

        # Всего 1 блок VSSD (вместо 4 стадий)
        self.vssd_block = nn.Sequential(
            # Local Perception
            nn.Conv2d(dim, dim, kernel_size=3, padding=1, groups=dim),
            nn.GELU(),

            # Global State Pooling
            nn.AdaptiveAvgPool2d(1),

            # Проекция
            nn.Conv2d(dim, dim * 2, kernel_size=1),  # уменьшили коэффициент
            nn.GELU(),
            nn.Conv2d(dim * 2, dim, kernel_size=1),
        )

        # Классификатор
        self.head = nn.Linear(dim, num_classes)

    def forward(self, x):
        # Патч эмбеддинг
        x = self.patch_embed(x)

        # VSSD блок
        identity = x
        x = self.vssd_block(x)
        x = x + identity  # residual

        # Классификация
        x = x.mean(dim=[2, 3])  # global pooling
        x = self.head(x)

        return x

# ==================== БАЗОВАЯ МОДЕЛЬ (CPU-ОПТИМИЗИРОВАНА) ====================
class LiteCNN(nn.Module):
    """Простая CNN для сравнения на CPU"""
    def __init__(self, num_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),  # уменьшили каналы
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.AdaptiveAvgPool2d(1)
        )
        self.classifier = nn.Linear(32, num_classes)  # уменьшили размерность

    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x

# ==================== ОБУЧЕНИЕ (ОПТИМИЗИРОВАНО ДЛЯ CPU) ====================
def train_model_lite(model, train_loader, test_loader, model_name, epochs=10):  # уменьшили эпохи
    """Быстрое обучение на CPU"""
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=1e-3)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=2, gamma=0.5)

    history = {'train_loss': [], 'train_acc': [], 'test_acc': []}

    print(f"\nОбучение {model_name}...")
    for epoch in range(epochs):
        # Обучение
        model.train()
        total_loss = 0
        correct = 0
        total = 0

        # Используем tqdm для отслеживания прогресса
        pbar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{epochs}')
        for batch_idx, (inputs, targets) in enumerate(pbar):
            inputs, targets = inputs.to(config.device), targets.to(config.device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()

            # Обновляем прогресс-бар
            pbar.set_postfix({
                'Loss': f'{loss.item():.4f}',
                'Acc': f'{100.*correct/total:.1f}%'
            })

        train_acc = 100. * correct / total
        history['train_loss'].append(total_loss / len(train_loader))
        history['train_acc'].append(train_acc)

        # Тестирование (быстрое)
        test_acc = evaluate(model, test_loader)
        history['test_acc'].append(test_acc)

        scheduler.step()

        print(f'  Train Acc: {train_acc:.2f}%, Test Acc: {test_acc:.2f}%, LR: {scheduler.get_last_lr()[0]:.6f}')

    return history, test_acc

def evaluate(model, test_loader):
    """Быстрая оценка"""
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for inputs, targets in test_loader:
            inputs, targets = inputs.to(config.device), targets.to(config.device)
            outputs = model(inputs)
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()

    return 100. * correct / total

# ==================== ВИЗУАЛИЗАЦИИ ====================
def plot_results_lite(history_vssd, history_cnn, model_names=["VSSD", "CNN"]):
    """Минимальные графики"""
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    # Точность на тесте
    axes[0].plot(history_vssd['test_acc'], 'o-', label=model_names[0], linewidth=2, markersize=8)
    axes[0].plot(history_cnn['test_acc'], 's-', label=model_names[1], linewidth=2, markersize=8)
    axes[0].set_xlabel('Эпоха')
    axes[0].set_ylabel('Точность (%)')
    axes[0].set_title('Сравнение точности на тесте')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)

    # Потери при обучении
    axes[1].plot(history_vssd['train_loss'], 'o-', label=model_names[0], linewidth=2, markersize=8)
    axes[1].plot(history_cnn['train_loss'], 's-', label=model_names[1], linewidth=2, markersize=8)
    axes[1].set_xlabel('Эпоха')
    axes[1].set_ylabel('Потери')
    axes[1].set_title('Сравнение потерь при обучении')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)

    # Точность при обучении
    axes[2].plot(history_vssd['train_acc'], 'o-', label=model_names[0], linewidth=2, markersize=8)
    axes[2].plot(history_cnn['train_acc'], 's-', label=model_names[1], linewidth=2, markersize=8)
    axes[2].set_xlabel('Эпоха')
    axes[2].set_ylabel('Точность (%)')
    axes[2].set_title('Точность при обучении')
    axes[2].legend()
    axes[2].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig('lite_results.png', dpi=150, bbox_inches='tight')
    plt.show()

def visualize_sample_predictions(model, test_loader, class_names=None):
    """Показываем несколько предсказаний"""
    model.eval()
    dataiter = iter(test_loader)
    images, labels = next(dataiter)

    if class_names is None and config.dataset == "CIFAR10":
        class_names = ['самолет', 'автомобиль', 'птица', 'кот', 'олень',
                       'собака', 'лягушка', 'лошадь', 'корабль', 'грузовик']
    else:
        class_names = [str(i) for i in range(config.num_classes)]

    fig, axes = plt.subplots(2, 5, figsize=(15, 6))
    axes = axes.ravel()

    with torch.no_grad():
        for i in range(10):
            if i >= len(images):
                break

            img = images[i].unsqueeze(0).to(config.device)
            output = model(img)
            _, predicted = output.max(1)

            # Денормализация
            img_np = images[i].permute(1, 2, 0).numpy()
            img_np = img_np * 0.5 + 0.5  # обратная нормализация
            img_np = np.clip(img_np, 0, 1)

            color = 'green' if predicted.item() == labels[i].item() else 'red'

            axes[i].imshow(img_np)
            axes[i].set_title(f'True: {class_names[labels[i]]}\nPred: {class_names[predicted.item()]}',
                            color=color, fontsize=10)
            axes[i].axis('off')

    plt.tight_layout()
    plt.savefig('predictions.png', dpi=150, bbox_inches='tight')
    plt.show()

# ==================== ОСНОВНОЙ ЭКСПЕРИМЕНТ (CPU) ====================
def main_lite():
    """Главная функция lite-версии для CPU"""
    print("="*60)
    print("LITE ЭКСПЕРИМЕНТ VSSD (CPU-версия)")
    print("="*60)
    print(f"Время выполнения: ~30-45 минут на CPU")
    print(f"Данные: {config.dataset} (подмножество из {config.subset_size} изображений)")
    print(f"Эпохи: {config.num_epochs}")
    print(f"Batch size: {config.batch_size}")
    print()

    start_time = time.time()

    # 1. Данные
    print("1. Загрузка данных...")
    train_dataset, test_dataset = get_lite_datasets()

    train_loader = DataLoader(
        train_dataset, batch_size=config.batch_size, shuffle=True, num_workers=0  # 0 для CPU
    )
    test_loader = DataLoader(
        test_dataset, batch_size=config.batch_size, shuffle=False, num_workers=0
    )

    print(f"   Обучающие: {len(train_dataset)} изображений")
    print(f"   Тестовые: {len(test_dataset)} изображений")

    # 2. Инициализация моделей
    print("\n2. Создание моделей...")
    vssd_model = LiteVSSD(num_classes=config.num_classes).to(config.device)
    cnn_model = LiteCNN(num_classes=config.num_classes).to(config.device)

    vssd_params = sum(p.numel() for p in vssd_model.parameters())
    cnn_params = sum(p.numel() for p in cnn_model.parameters())

    print(f"   VSSD параметры: {vssd_params:,}")
    print(f"   CNN параметры: {cnn_params:,}")
    print(f"   Отношение параметров VSSD/CNN: {vssd_params/cnn_params:.2f}x")

    # 3. Обучение VSSD
    print("\n3. Обучение VSSD...")
    history_vssd, final_acc_vssd = train_model_lite(
        vssd_model, train_loader, test_loader,
        "VSSD", epochs=config.num_epochs
    )

    # 4. Обучение CNN
    print("\n4. Обучение CNN (аналог)...")
    history_cnn, final_acc_cnn = train_model_lite(
        cnn_model, train_loader, test_loader,
        "CNN", epochs=config.num_epochs
    )

    # 5. Визуализации
    print("\n5. Визуализация результатов...")
    plot_results_lite(history_vssd, history_cnn)

    # Показать примеры предсказаний
    print("\nПримеры предсказаний VSSD:")
    visualize_sample_predictions(vssd_model, test_loader)

    # 6. Результаты
    total_time = time.time() - start_time

    print("\n" + "="*60)
    print("РЕЗУЛЬТАТЫ LITE ЭКСПЕРИМЕНТА")
    print("="*60)

    improvement = ((final_acc_vssd - final_acc_cnn) / final_acc_cnn) * 100

    print(f"\nФинальная точность:")
    print(f"  VSSD: {final_acc_vssd:.2f}%")
    print(f"  CNN:  {final_acc_cnn:.2f}%")
    print(f"  Улучшение: {improvement:+.2f}%")

    print(f"\nПараметры моделей:")
    print(f"  VSSD: {vssd_params:,}")
    print(f"  CNN:  {cnn_params:,}")
    print(f"  VSSD больше CNN в {vssd_params/cnn_params:.2f} раз")

    print(f"\nОбщее время: {total_time/60:.1f} минут")

    # 7. Выводы
    print("\n" + "="*60)
    print("ВЫВОДЫ ДЛЯ ОТЧЕТА")
    print("="*60)

    if improvement > 0:
        print("✓ VSSD показывает лучшую точность на маленьком датасете")
        print(f"✓ Улучшение точности: {improvement:+.2f}%")
        print("✓ Non-causal механизм VSSD работает эффективно")
    else:
        print(f"✗ VSSD уступает CNN на {abs(improvement):.2f}%")
        print("✗ На маленьком датасете преимущества VSSD могут не проявляться")
        print("✗ Возможно, требуется больше данных или эпох обучения")

    print("\nКлючевые наблюдения:")
    print("1. VSSD имеет более сложную архитектуру с глобальным контекстом")
    print("2. CNN быстрее обучается благодаря локальным сверткам")
    print("3. VSSD потенциально лучше масштабируется на больших данных")
    print(f"4. Время эксперимента: {total_time/60:.1f} минут на {config.device}")

    # Сохраняем результаты
    results = {
        'VSSD': {
            'accuracy': float(final_acc_vssd),
            'params': int(vssd_params),
            'train_loss': [float(x) for x in history_vssd['train_loss']],
            'test_acc': [float(x) for x in history_vssd['test_acc']]
        },
        'CNN': {
            'accuracy': float(final_acc_cnn),
            'params': int(cnn_params),
            'train_loss': [float(x) for x in history_cnn['train_loss']],
            'test_acc': [float(x) for x in history_cnn['test_acc']]
        },
        'improvement': float(improvement),
        'time_minutes': float(total_time/60),
        'config': {
            'dataset': config.dataset,
            'subset_size': config.subset_size,
            'epochs': config.num_epochs,
            'batch_size': config.batch_size,
            'device': str(config.device)
        }
    }

    # Сохранение в файл
    import json
    with open('lite_experiment_results.json', 'w') as f:
        json.dump(results, f, indent=2)

    with open('lite_experiment_summary.txt', 'w') as f:
        f.write("="*50 + "\n")
        f.write("РЕЗУЛЬТАТЫ ЭКСПЕРИМЕНТА VSSD vs CNN\n")
        f.write("="*50 + "\n\n")
        f.write(f"Датасет: {config.dataset}\n")
        f.write(f"Изображений: {config.subset_size}\n")
        f.write(f"Эпохи: {config.num_epochs}\n")
        f.write(f"Устройство: {config.device}\n\n")

        f.write(f"{'Модель':<10} {'Точность':<10} {'Параметры':<12} {'Улучшение':<10}\n")
        f.write("-"*45 + "\n")
        f.write(f"{'VSSD':<10} {final_acc_vssd:<10.2f}% {vssd_params:<12,} {'-':<10}\n")
        f.write(f"{'CNN':<10} {final_acc_cnn:<10.2f}% {cnn_params:<12,} {improvement:+.2f}%\n\n")

        f.write(f"Время выполнения: {total_time/60:.1f} минут\n")

        if improvement > 0:
            f.write(f"\n✓ VSSD превосходит CNN на {improvement:.2f}%\n")
        else:
            f.write(f"\n✗ VSSD уступает CNN на {abs(improvement):.2f}%\n")

    print(f"\n✓ Результаты сохранены в файлы:")
    print(f"  - lite_experiment_results.json (детальные)")
    print(f"  - lite_experiment_summary.txt (краткие)")
    print(f"  - lite_results.png (графики)")
    print(f"  - predictions.png (примеры предсказаний)")

    return results

# ==================== ЗАПУСК В COLAB ====================
if __name__ == "__main__":
    print("Запуск lite-эксперимента VSSD...")
    print("="*60)

    # Инструкция для Colab
    if str(config.device) == 'cpu':
        print("\nВАЖНО: Вы используете CPU, что будет медленнее.")
        print("Для ускорения в Google Colab:")
        print("1. В меню выберите 'Runtime' → 'Change runtime type'")
        print("2. В 'Hardware accelerator' выберите 'GPU'")
        print("3. Нажмите 'Save' и перезапустите ноутбук\n")

    results = main_lite()

    print("\n" + "="*60)
    print("ЭКСПЕРИМЕНТ ЗАВЕРШЕН!")
    print("="*60)
```
